# Austin-San Marcos Housing Affordability Analysis

## Project Proposal

This project analyzes housing affordability in the Austin to San Marcos corridor using three complementary datasets: Redfin market data (median sale price, inventory, days on market, sale-to-list ratio), Zillow Home Value Index trends at the city level, and the City of Austin's Affordable Housing Inventory with ZIP-code-level coordinates. By combining temporal trend analysis of housing prices and market dynamics with geospatial visualization of where affordable housing units are concentrated across Austin ZIP codes, the project will examine how affordability has shifted over time in both cities, whether San Marcos has maintained a meaningful price advantage over Austin, and where gaps in affordable housing supply are most acute. The income-needed metric from Zillow will further contextualize these trends by estimating the household income required to purchase a home in the Austin metro, connecting price movements to real purchasing power.
